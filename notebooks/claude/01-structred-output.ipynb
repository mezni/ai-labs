```python
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()

client = Anthropic()
model = "claude-haiku-4-5-20251001"

message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[{"role": "user", "content": "Say hello in one word"}],
)
print(message.content[0].text)
```

```python
from pydantic import BaseModel, Field


class MovieReview(BaseModel):
    title: str = Field(description="The title of the movie")
    director: str = Field(description="The director of the movie")
    rating: int = Field(description="Rating from 1 to 10")
    genre: list[str] = Field(description="List of genres")
    summary: str = Field(description="A brief 1-2 sentence summary")


response = client.messages.parse(
    model=model,
    max_tokens=1000,
    output_format=MovieReview,
    messages=[
        {
            "role": "user",
            "content": "Give me a review of the movie Inception directed by Christopher Nolan.",
        }
    ],
)

review = response.parsed_output
print(f"Title:    {review.title}")
print(f"Director: {review.director}")
print(f"Rating:   {review.rating}/10")
print(f"Genres:   {', '.join(review.genre)}")
print(f"Summary:  {review.summary}")
```
